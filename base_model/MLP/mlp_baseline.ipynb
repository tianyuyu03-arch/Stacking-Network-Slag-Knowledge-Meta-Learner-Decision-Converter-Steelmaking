```python
import pandas as pd
import numpy as np
from sklearn.neural_network import MLPRegressor
from sklearn.multioutput import MultiOutputRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")

# 1. 加载数据
print("正在加载数据...")
# 修改数据路径与LightGBM保持一致
df_original = pd.read_excel('../../data/pinghua_data_x_names.xlsx')
print(f"原始数据集大小: {df_original.shape}")

# 2. 定义目标变量 - 与LightGBM保持一致
target_vars = ['x25', 'x26', 'x27', 'x28', 'x29']

# 3. 分离特征和目标
X_original = df_original.drop(target_vars, axis=1)
y_original = df_original[target_vars]

# 4. 分割原始数据为训练集和测试集
X_train, X_test, y_train, y_test = train_test_split(
    X_original, y_original, test_size=0.2, random_state=42
)

print(f"训练集大小: {X_train.shape}")
print(f"测试集大小: {X_test.shape}")

# 5. 数据标准化
scaler_X = StandardScaler()
X_train_scaled = scaler_X.fit_transform(X_train)
X_test_scaled = scaler_X.transform(X_test)
# 最优参数:
# hidden_layer_sizes: (np.int64(103), np.int64(16))
# activation: tanh
# alpha: 0.01
# learning_rate_init: 0.1
# learning_rate: adaptive
# early_stopping: False
# n_iter_no_change: 19
# max_iter: 1000
# solver: adam
# random_state: 42
# validation_fraction: 0.1
# tol: 0.0001
# batch_size: auto
# 6. 创建并训练MLP模型
print("\n开始训练MLP神经网络模型...")
model = MultiOutputRegressor(MLPRegressor(
    hidden_layer_sizes=(103, 16),  # 优化后的隐藏层结构
    activation='tanh',            # 优化后的激活函数
    solver='adam',               # 优化器保持不变
    alpha=0.01,                  # 优化后的L2正则化参数
    batch_size='auto',           # 批量大小保持不变
    learning_rate='adaptive',    # 学习率类型保持不变
    learning_rate_init=0.1,      # 优化后的初始学习率
    max_iter=1000,               # 最大迭代次数保持不变
    tol=0.0001,                  # 收敛容差保持不变
    early_stopping=False,        # 优化后关闭提前停止
    validation_fraction=0.1,     # 验证比例保持不变
    n_iter_no_change=19,         # 优化后的无改善等待迭代数
    random_state=42              # 随机种子保持不变
))

# 使用训练数据训练
model.fit(X_train_scaled, y_train)
print("模型训练完成！")

# 7. 在测试集上进行预测
y_pred = model.predict(X_test_scaled)

# 8. 评估模型性能
print("\n模型评估指标:")
overall_r2 = 0
overall_rmse = 0
overall_mae = 0

for idx, name in enumerate(target_vars):
    # 计算指标
    mse = mean_squared_error(y_test.iloc[:, idx], y_pred[:, idx])
    rmse = np.sqrt(mse)
    mae = mean_absolute_error(y_test.iloc[:, idx], y_pred[:, idx])
    r2 = r2_score(y_test.iloc[:, idx], y_pred[:, idx])
    
    # 累加指标值
    overall_r2 += r2
    overall_rmse += rmse
    overall_mae += mae
    
    # 打印每个目标变量的指标
    print(f"\n{name} 指标:")
    print(f"MSE={mse:.4f}, RMSE={rmse:.4f}, MAE={mae:.4f}, R²={r2:.4f}")

# 计算平均指标
overall_r2 /= len(target_vars)
overall_rmse /= len(target_vars)
overall_mae /= len(target_vars)

print("\n总体平均指标:")
print(f"平均 RMSE: {overall_rmse:.4f}")
print(f"平均 MAE: {overall_mae:.4f}")
print(f"平均 R²: {overall_r2:.4f}")
```
